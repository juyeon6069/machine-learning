```python
import tensorflow as tf
import numpy as np
import random
```

```python
# Hyperparameters
learning_rate = 0.001
batch_size = 100
training_epochs = 15
nb_classes = 10

random.seed(777)  # for reproducibility

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Original shape:", x_train.shape)

# Flatten images (28x28 -> 784)
x_train = x_train.reshape(-1, 28 * 28)
x_test = x_test.reshape(-1, 28 * 28)

# Normalize (0~255 -> 0~1 Normalize)
x_train = x_train / 255.0
x_test = x_test / 255.0

# One-hot encode labels
y_train = tf.keras.utils.to_categorical(y_train, nb_classes)
y_test = tf.keras.utils.to_categorical(y_test, nb_classes)
```

```python
# Simple 1-Layer NN (10 units, softmax)

model1 = tf.keras.Sequential([
    tf.keras.layers.Dense(units=10, input_dim=784, activation='softmax')
])

model1.compile(loss='categorical_crossentropy',
               optimizer=tf.keras.optimizers.Adam(learning_rate),
               metrics=['accuracy'])

model1.summary()
history1 = model1.fit(x_train, y_train, batch_size=batch_size, epochs=training_epochs, verbose=0)
loss, acc = model1.evaluate(x_test, y_test)
print("Test accuracy:", acc)
```

```python
# Basic NN (256 → 256 → 10 softmax)

model = tf.keras.Sequential([
    tf.keras.layers.Dense(256, input_dim=784, kernel_initializer='glorot_normal', activation='relu'),
    tf.keras.layers.Dense(256, kernel_initializer='glorot_normal', activation='relu'),
    tf.keras.layers.Dense(nb_classes, kernel_initializer='glorot_normal', activation='softmax')
])

model.compile(
    loss='categorical_crossentropy',
    optimizer=tf.keras.optimizers.Adam(learning_rate),
    metrics=['accuracy']
)

model.summary()

history = model.fit(x_train, y_train, batch_size=batch_size, epochs=training_epochs, verbose=0)

# Evaluate
loss, acc = model.evaluate(x_test, y_test)
print("Test accuracy:", acc)
```

```python
# Deep NN (512 → 512 → 512 → 512 → 10 softmax)
# Xavier(Glorot) initialization
    # 깊은 네트워크에서 Vanishing & Exploding Gradient 문제를 줄여줌 
    # => 안정적인 학습을 가능하게 함

model2 = tf.keras.Sequential([
    tf.keras.layers.Dense(512, input_dim=784, kernel_initializer='glorot_normal', activation='relu'),
    tf.keras.layers.Dense(512, kernel_initializer='glorot_normal', activation='relu'),
    tf.keras.layers.Dense(512, kernel_initializer='glorot_normal', activation='relu'),
    tf.keras.layers.Dense(512, kernel_initializer='glorot_normal', activation='relu'),
    tf.keras.layers.Dense(nb_classes, kernel_initializer='glorot_normal', activation='softmax')
])

model2.compile(
    loss='categorical_crossentropy',
    optimizer=tf.keras.optimizers.Adam(learning_rate),
    metrics=['accuracy']
)

model2.summary()

history2 = model2.fit(x_train, y_train, batch_size=batch_size, epochs=training_epochs, verbose=0)

# Evaluate
loss, acc = model2.evaluate(x_test, y_test)
print("Test accuracy:", acc)
```

```python
# Dropout NN (512 → Dropout → 512 → Dropout …)
    # Dropout으로 일부 뉴런을 랜덤 비활성화 -> 특정 feature에 과도하게 의존X
    # => overfitting 방지

drop_rate = 0.3

model3 = tf.keras.Sequential([
    tf.keras.layers.Dense(512, input_dim=784, activation='relu'),
    tf.keras.layers.Dropout(drop_rate),

    tf.keras.layers.Dense(512, activation='relu'),
    tf.keras.layers.Dropout(drop_rate),

    tf.keras.layers.Dense(512, activation='relu'),
    tf.keras.layers.Dropout(drop_rate),

    tf.keras.layers.Dense(nb_classes, activation='softmax')
])

model3.compile(
    loss='categorical_crossentropy',
    optimizer=tf.keras.optimizers.Adam(learning_rate),
    metrics=['accuracy']
)

model3.summary()

history3 = model3.fit(x_train, y_train, batch_size=batch_size, epochs=training_epochs, verbose=0)

# Evaluate
loss, acc = model3.evaluate(x_test, y_test)
print("Test accuracy:", acc)
```

```python
# Random Predictions (for all models)

def show_random_predictions(model, x_test, y_test, n=10):
    y_pred = model.predict(x_test)
    for _ in range(n):
        idx = random.randint(0, x_test.shape[0]-1)
        print(f"Index: {idx}, Actual: {np.argmax(y_test[idx])}, Predicted: {np.argmax(y_pred[idx])}")

print("\nBasic NN Predictions:")
show_random_predictions(model, x_test, y_test)

print("\nDeep NN Predictions:")
show_random_predictions(model2, x_test, y_test)

print("\nDropout NN Predictions:")
show_random_predictions(model3, x_test, y_test)
```

```python
import tensorflow as tf
import numpy as np

# 데이터 
    # 입력 feature 2개(x1,x2)
    # 출력 0 또는 1
x_data = np.array([[1, 2],
                   [2, 3],
                   [3, 1],
                   [4, 3],
                   [5, 3],
                   [6, 2]], dtype=float)
y_data = np.array([[0],
                   [0],
                   [0],
                   [1],
                   [1],
                   [1]], dtype=float)

# 모델 정의
# 'sigmoid' → 출력 확률로 변환: 0~1(선형 합을 0~1 확률로 압축)
model = tf.keras.Sequential()
model.add(tf.keras.layers.Dense(units=1, input_dim=2, activation='sigmoid'))

# 컴파일
    # loss='binary_crossentropy'
    # optimizer=SGD(0.01)
    # metrics=['accuracy']
model.compile(loss='binary_crossentropy',
              optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
              metrics=['accuracy'])

# 모델 요약
model.summary()

# 학습
history = model.fit(x_data, y_data, epochs=500, verbose=0)

# Accuracy 출력
print("Accuracy: ", history.history['accuracy'][-1])
```

```python
# diabetes

import tensorflow as tf
import numpy as np

# CSV 데이터 로드
xy = np.loadtxt('data-03-diabetes.csv', delimiter=',', dtype=np.float32)
x_data = xy[:, 0:-1]
y_data = xy[:, [-1]]

print(x_data.shape, y_data.shape)  # 확인

# 모델 정의
model = tf.keras.Sequential()
model.add(tf.keras.layers.Dense(units=1, input_dim=x_data.shape[1], activation='sigmoid'))

# 컴파일
model.compile(loss='binary_crossentropy',
              optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
              metrics=['accuracy'])

model.summary()

# 학습
history = model.fit(x_data, y_data, epochs=500, verbose=0)

# Accuracy 출력
print("Accuracy: {0}".format(history.history['accuracy'][-1]))

# 단일 데이터 예측
y_predict = model.predict(np.array([[0.176471, 0.155779, 0, 0, 0, 0.052161, -0.952178, -0.733333]]))
print("Predict: {0}".format(y_predict))

# 전체 데이터 평가
evaluate = model.evaluate(np.array(x_data), np.array(y_data))
print("loss: {0}".format(evaluate[0]))
```

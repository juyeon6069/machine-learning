```python
import numpy as np
import tensorflow as tf

# 데이터
x_train = np.array([1, 2, 3, 4], dtype=float)
y_train = np.array([0, -1, -2, -3], dtype=float)

# 모델 정의
# Dense layer : y = W1*x1 + W2*x2 + ... + Wn*xn + b
    # W1~Wn = 각 입력에 곱하는 가중치(weight)
    # b = 편향(bias) (모든 입력에 공통으로 더하는 값)
# Dense layer: multi variable → W n개 => deep learning → 이런 층이 수십~수백 개 연결됨
model = tf.keras.Sequential()
model.add(tf.keras.layers.Dense(units=1, input_dim=1))

# 옵티마이저와 손실
sgd = tf.keras.optimizers.SGD(learning_rate=0.1)
model.compile(loss='mse', optimizer=sgd)

# 모델 요약
model.summary()

# 학습
model.fit(x_train, y_train, epochs=200, verbose=0)

# 예측
y_predict = model.predict(np.array([5, 4]))
print(y_predict)
```

```python
# Multi-variable linear regression
import tensorflow as tf
import numpy as np

x_data = np.array([[73., 80., 75.],
                   [93., 88., 93.],
                   [89., 91., 90.],
                   [96., 98., 100.],
                   [73., 66., 70.]], dtype=float)

y_data = np.array([[152.],
                   [185.],
                   [180.],
                   [196.],
                   [142.]], dtype=float)

# 입력이 여러개 => y = W1*x1 + W2*x2 + W3*x3 + b (학습해야 할 매개변수가 W 3개 + b 1개 = 4개)
model = tf.keras.Sequential()
model.add(tf.keras.layers.Dense(units=1, input_dim=3)) 

# learning rate를 작게해야하는 이유
    # 입력 숫자가 크고 변동이 크기 때문
    # learning rate를 크게 하면 loss가 발산함
model.compile(loss='mse', optimizer=tf.keras.optimizers.SGD(learning_rate=1e-5))

model.summary()

history = model.fit(x_data, y_data, epochs=100, verbose=0)

y_predict = model.predict(np.array([[72., 93., 90.]]))
print("Predicted value:", y_predict)

```

```python
# Multi-variable linear regression(file_input)
import tensorflow as tf
import numpy as np

# CSV 데이터 로드
xy = np.loadtxt('data-01-test-score.csv', delimiter=',', dtype=np.float32)
x_data = xy[:, 0:-1] # 입력 3개(마지막 열 제외)
y_data = xy[:, [-1]] # 출력 1개(마지막 열만 가져오기)

# print(x_data, "\nx_data shape:", x_data.shape) # x_data: (N, 3)
# print(y_data, "\ny_data shape:", y_data.shape) # y_data: (N, 1)

model = tf.keras.Sequential()
model.add(tf.keras.layers.Dense(units=1, input_dim=3))  # linear activation 기본값

model.compile(loss='mse', optimizer=tf.keras.optimizers.SGD(learning_rate=1e-5))

history = model.fit(x_data, y_data, epochs=2000, verbose=0)

# 예측 = (가중치 * 입력값)의 합 + 편향
print("Your score will be:", model.predict(np.array([[100, 70, 101]])))
print("Other scores will be:", model.predict(np.array([[60, 70, 110], [90, 100, 80]])))
```
